```python

# ============================================================
# EXPLAINABLE MANUFACTURING PROCESS PATTERN ANALYSIS
# AND ADAPTIVE RISK MONITORING
#
# Application: Injection-molded automotive plastic components
# Validation: Real plastic road-lens manufacturing dataset
#
# Algorithms:
# 1. K-Means
# 2. DBSCAN
# 3. Gaussian Mixture Model (GMM)
# 4. Random Forest (explanation surrogate)
# 5. Kolmogorov-Smirnov Process Drift Detection
#
# Dataset source:
# https://github.com/airtlab/
# machine-learning-for-quality-prediction-in-plastic-injection-molding
# ============================================================

!pip -q install pandas numpy scikit-learn matplotlib seaborn joblib

import os
import json
import joblib
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, DBSCAN
from sklearn.mixture import GaussianMixture
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score,
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score
)
from sklearn.ensemble import RandomForestClassifier
from scipy.stats import ks_2samp

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

print("=" * 65)
print("MANUFACTURING PROCESS PATTERN ANALYSIS")
print("=" * 65)


# ============================================================
# STEP 1: LOAD REAL-WORLD MANUFACTURING DATASET
# ============================================================

DATA_URL = (
    "https://raw.githubusercontent.com/"
    "airtlab/machine-learning-for-quality-prediction-in-plastic-injection-molding/"
    "/content/verified_real_world_plastic_injection_dataset_extract.csv"
)

def load_dataset():
    try:
        df = pd.read_csv(DATA_URL, sep=None, engine="python")
        print("Dataset downloaded successfully.")
        return df

    except Exception as e:
        print("Automatic download failed:", e)
        print("Please upload the original CSV.")

        from google.colab import files
        uploaded = files.upload()

        if not uploaded:
            raise ValueError("No dataset uploaded.")

        filename = next(iter(uploaded))
        df = pd.read_csv(filename, sep=None, engine="python")

        return df


df = load_dataset()

print("\nDataset shape:", df.shape)
print("\nFirst five records:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())


# ============================================================
# STEP 2: IDENTIFY QUALITY COLUMN
# ============================================================

df.columns = df.columns.str.strip()

quality_candidates = [
    c for c in df.columns
    if c.lower() in ["quality", "quality_class", "label"]
]

if not quality_candidates:
    raise ValueError(
        "Quality column not found. Check the uploaded dataset."
    )

quality_col = quality_candidates[0]

df[quality_col] = pd.to_numeric(
    df[quality_col],
    errors="coerce"
)

# Original dataset quality coding
quality_map = {
    1: "Waste",
    2: "Acceptable",
    3: "Target",
    4: "Inefficient"
}

df = df.dropna(subset=[quality_col]).copy()
df[quality_col] = df[quality_col].astype(int)

if not set(df[quality_col].unique()).issubset(quality_map):
    raise ValueError(
        "Unexpected quality labels. Verify dataset coding."
    )

df["Quality_Name"] = df[quality_col].map(quality_map)

# Waste = defective group for binary risk evaluation
df["Defect"] = (df[quality_col] == 1).astype(int)

df["Record_ID"] = np.arange(1, len(df) + 1)

print("\nQuality distribution:")
print(df["Quality_Name"].value_counts())

sns.countplot(
    data=df,
    x="Quality_Name",
    order=["Waste", "Acceptable", "Target", "Inefficient"]
)
plt.title("Actual Manufacturing Quality Distribution")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()


# ============================================================
# STEP 3: DATA CLEANING
# ============================================================

feature_cols = [
    c for c in df.columns
    if c not in [
        quality_col,
        "Quality_Name",
        "Defect",
        "Record_ID"
    ]
]

# Convert process measurements to numeric values
for col in feature_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Remove features with too many missing values
missing_ratio = df[feature_cols].isna().mean()

feature_cols = [
    c for c in feature_cols
    if missing_ratio[c] <= 0.40
]

# Remove constant features
feature_cols = [
    c for c in feature_cols
    if df[c].nunique(dropna=True) > 1
]

if len(feature_cols) < 2:
    raise ValueError("Insufficient useful numeric process features.")

print("\nFeatures selected:", len(feature_cols))
print(feature_cols)

print("\nMissing values:")
print(df[feature_cols].isnull().sum())


# ============================================================
# STEP 4: TRAIN / TEST SPLIT
#
# The model discovers patterns using training data.
# Held-out records simulate future production observations.
# ============================================================

train_idx, test_idx = train_test_split(
    df.index,
    test_size=0.25,
    stratify=df[quality_col],
    random_state=RANDOM_STATE
)

train_df = df.loc[train_idx].copy()
test_df = df.loc[test_idx].copy()

print("\nTraining samples:", len(train_df))
print("Testing samples:", len(test_df))


# ============================================================
# STEP 5: IMPUTATION AND SCALING
# ============================================================

imputer = SimpleImputer(strategy="median")

X_train_raw = imputer.fit_transform(
    train_df[feature_cols]
)

X_test_raw = imputer.transform(
    test_df[feature_cols]
)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_raw)
X_test_scaled = scaler.transform(X_test_raw)

print("\nPreprocessing completed.")


# ============================================================
# STEP 6: EXPLORATORY DATA ANALYSIS
# ============================================================

print("\nStatistical summary:")
display(train_df[feature_cols].describe())

plt.figure(figsize=(11, 8))

sns.heatmap(
    train_df[feature_cols].corr(),
    cmap="coolwarm",
    center=0
)

plt.title("Manufacturing Process Feature Correlation")
plt.tight_layout()
plt.show()


# ============================================================
# STEP 7: PCA DIMENSIONALITY REDUCTION
# ============================================================

pca = PCA(n_components=0.95, svd_solver="full")

X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)

print("\nOriginal features:", X_train_scaled.shape[1])
print("PCA components retained:", X_train_pca.shape[1])

plt.figure(figsize=(8, 4))

plt.plot(
    np.cumsum(pca.explained_variance_ratio_),
    marker="o"
)

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA Explained Variance")
plt.tight_layout()
plt.show()


# ============================================================
# STEP 8: COMPARE CLUSTERING ALGORITHMS
# ============================================================

results = []
model_candidates = []

def evaluate_clustering(name, model, labels, X):
    labels = np.asarray(labels)

    # DBSCAN noise (-1) excluded from internal metrics
    valid = labels != -1
    unique_clusters = np.unique(labels[valid])

    if (
        len(unique_clusters) < 2
        or valid.sum() <= len(unique_clusters)
    ):
        return

    X_valid = X[valid]
    labels_valid = labels[valid]

    sil = silhouette_score(X_valid, labels_valid)

    db = davies_bouldin_score(X_valid, labels_valid)

    ch = calinski_harabasz_score(
        X_valid, labels_valid
    )

    results.append({
        "Model": name,
        "Silhouette": sil,
        "Davies_Bouldin": db,
        "Calinski_Harabasz": ch,
        "Clusters": len(unique_clusters),
        "Noise_Ratio": 1 - valid.mean()
    })

    model_candidates.append({
        "name": name,
        "model": model,
        "labels": labels,
        "score": sil,
        "noise_ratio": 1 - valid.mean()
    })


# K-Means models
for k in range(2, 9):

    model = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=10
    )

    labels = model.fit_predict(X_train_pca)

    evaluate_clustering(
        f"KMeans_{k}",
        model,
        labels,
        X_train_pca
    )


# Gaussian Mixture Models
for k in range(2, 9):

    model = GaussianMixture(
        n_components=k,
        covariance_type="diag",
        reg_covar=1e-5,
        random_state=RANDOM_STATE
    )

    labels = model.fit_predict(X_train_pca)

    evaluate_clustering(
        f"GMM_{k}",
        model,
        labels,
        X_train_pca
    )


# DBSCAN
for eps in [0.7, 1.0, 1.5, 2.0, 3.0]:

    model = DBSCAN(
        eps=eps,
        min_samples=10
    )

    labels = model.fit_predict(X_train_pca)

    evaluate_clustering(
        f"DBSCAN_{eps}",
        model,
        labels,
        X_train_pca
    )


results_df = pd.DataFrame(results)

if results_df.empty:
    raise RuntimeError("No valid clustering solutions found.")

results_df = results_df.sort_values(
    "Silhouette",
    ascending=False
)

print("\nCLUSTERING MODEL COMPARISON:")
display(results_df)

plt.figure(figsize=(12, 5))

sns.barplot(
    data=results_df,
    x="Model",
    y="Silhouette"
)

plt.xticks(rotation=75)
plt.title("Clustering Model Comparison")
plt.tight_layout()
plt.show()


# ============================================================
# STEP 9: SELECT BEST DEPLOYABLE MODEL
#
# DBSCAN has no native prediction method.
# For this prototype, deploy the highest-scoring
# K-Means/GMM model and retain DBSCAN for comparison.
# Selection uses clustering metrics, NOT quality labels.
# ============================================================

deployable = [
    item for item in model_candidates
    if item["name"].startswith(("KMeans", "GMM"))
]

best = max(
    deployable,
    key=lambda item: item["score"]
)

best_model = best["model"]
best_name = best["name"]

train_labels = best["labels"]
test_labels = best_model.predict(X_test_pca)

train_df["Cluster"] = train_labels
test_df["Cluster"] = test_labels

print("\nBEST DEPLOYABLE MODEL:", best_name)
print("Silhouette Score:", round(best["score"], 4))


# ============================================================
# STEP 10: VISUALIZE PRODUCTION PATTERNS
# ============================================================

pca_visual = PCA(n_components=2)

visual_2d = pca_visual.fit_transform(X_train_scaled)

plt.figure(figsize=(9, 6))

sns.scatterplot(
    x=visual_2d[:, 0],
    y=visual_2d[:, 1],
    hue=train_labels,
    palette="tab10",
    s=35
)

plt.title("Manufacturing Production Patterns")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.tight_layout()
plt.show()


# ============================================================
# STEP 11: QUALITY AND EFFICIENCY PROFILING
# ============================================================

cluster_profile = train_df.groupby("Cluster").agg(
    Samples=("Defect", "size"),
    Defects=("Defect", "sum"),
    Defect_Rate=("Defect", "mean")
)

# An inefficient quality category is not necessarily an
# actual measured energy inefficiency.
# Here it is the dataset's original quality class.

inefficient_rate = train_df.groupby("Cluster")[
    "Quality_Name"
].apply(lambda x: (x == "Inefficient").mean())

cluster_profile["Inefficient_Rate"] = inefficient_rate

cluster_profile["Defect_Risk_Lift"] = (
    cluster_profile["Defect_Rate"]
    / max(train_df["Defect"].mean(), 1e-9)
)

print("\nCLUSTER RISK PROFILE:")
display(cluster_profile)

quality_by_cluster = pd.crosstab(
    train_df["Cluster"],
    train_df["Quality_Name"],
    normalize="index"
)

quality_by_cluster.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 5)
)

plt.title("Quality Distribution Across Manufacturing Patterns")
plt.ylabel("Proportion")
plt.xlabel("Cluster")
plt.legend(title="Actual QC")
plt.tight_layout()
plt.show()


# ============================================================
# STEP 12: ASSIGN RISK LEVELS
#
# Risk labels are calibrated using training QC results.
# Thresholds are illustrative and should be validated
# for actual deployment.
# ============================================================

overall_defect_rate = train_df["Defect"].mean()

def get_risk_level(row):

    if row["Defect_Rate"] >= 2 * overall_defect_rate:
        return "HIGH"

    elif row["Defect_Rate"] >= overall_defect_rate:
        return "MEDIUM"

    else:
        return "LOW"


cluster_profile["Risk_Level"] = cluster_profile.apply(
    get_risk_level,
    axis=1
)

risk_map = cluster_profile["Risk_Level"].to_dict()
defect_rate_map = cluster_profile["Defect_Rate"].to_dict()

train_df["Risk_Level"] = train_df["Cluster"].map(risk_map)
test_df["Risk_Level"] = test_df["Cluster"].map(risk_map)

print("\nCLUSTER RISK ASSESSMENT:")
display(cluster_profile)


# ============================================================
# STEP 13: EXPLAIN CLUSTER CHARACTERISTICS
#
# Compare each cluster's original-process feature mean
# with the overall training mean in standardized units.
#
# This identifies association, NOT causation.
# ============================================================

explanation_rows = []

overall_mean = X_train_scaled.mean(axis=0)

for cluster_id in sorted(np.unique(train_labels)):

    mask = train_labels == cluster_id

    cluster_mean = X_train_scaled[mask].mean(axis=0)

    deviation = cluster_mean - overall_mean

    indices = np.argsort(
        np.abs(deviation)
    )[::-1][:5]

    for rank, idx in enumerate(indices, start=1):

        explanation_rows.append({
            "Cluster": int(cluster_id),
            "Rank": rank,
            "Feature": feature_cols[idx],
            "Standardized_Deviation": float(deviation[idx]),
            "Direction": (
                "Higher" if deviation[idx] > 0 else "Lower"
            )
        })


explanation_df = pd.DataFrame(explanation_rows)

print("\nTOP PROCESS VARIABLES PER CLUSTER:")
display(explanation_df)


# ============================================================
# STEP 14: EXPLANATION SURROGATE MODEL
#
# This model learns to reproduce clustering decisions.
# Its feature importance explains cluster assignments,
# not proven defect causes.
# ============================================================

surrogate = RandomForestClassifier(
    n_estimators=200,
    random_state=RANDOM_STATE,
    class_weight="balanced"
)

surrogate.fit(X_train_scaled, train_labels)

importance_df = pd.DataFrame({
    "Feature": feature_cols,
    "Importance": surrogate.feature_importances_
}).sort_values("Importance", ascending=False)

print("\nGLOBAL CLUSTER FEATURE IMPORTANCE:")
display(importance_df.head(10))

plt.figure(figsize=(10, 6))

sns.barplot(
    data=importance_df.head(10),
    x="Importance",
    y="Feature"
)

plt.title("Variables Explaining Manufacturing Cluster Membership")
plt.tight_layout()
plt.show()


# ============================================================
# STEP 15: TEST-SET RISK EVALUATION
#
# Risk probabilities come from training cluster statistics.
# Held-out quality labels are used only for evaluation.
# ============================================================

test_df["Predicted_Defect_Probability"] = (
    test_df["Cluster"].map(defect_rate_map)
)

test_df["Predicted_Defect"] = (
    test_df["Risk_Level"] == "HIGH"
).astype(int)

y_true = test_df["Defect"]
y_pred = test_df["Predicted_Defect"]

print("\nRISK ALERT CLASSIFICATION REPORT:")
print(
    classification_report(
        y_true,
        y_pred,
        zero_division=0
    )
)

print("Precision:", precision_score(
    y_true, y_pred, zero_division=0
))

print("Recall:", recall_score(
    y_true, y_pred, zero_division=0
))

print("F1 Score:", f1_score(
    y_true, y_pred, zero_division=0
))

print("PR-AUC (Average Precision):",
      average_precision_score(
          y_true,
          test_df["Predicted_Defect_Probability"]
      ))

cm = confusion_matrix(y_true, y_pred)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Greys"
)

plt.title("High-Risk Alert Confusion Matrix")
plt.xlabel("Predicted Defect Alert")
plt.ylabel("Actual Defect")
plt.tight_layout()
plt.show()


# ============================================================
# STEP 16: NEW PRODUCTION RECORD MONITORING
#
# Simulated live monitoring:
# A held-out production record is passed through the
# already-trained preprocessing and clustering pipeline.
# ============================================================

def monitor_new_record(record):

    record_df = pd.DataFrame([record])

    record_df = record_df.reindex(columns=feature_cols)

    for col in feature_cols:
        record_df[col] = pd.to_numeric(
            record_df[col],
            errors="coerce"
        )

    processed = imputer.transform(record_df)
    scaled = scaler.transform(processed)
    transformed = pca.transform(scaled)

    cluster_id = int(
        best_model.predict(transformed)[0]
    )

    risk_level = risk_map[cluster_id]

    defect_probability = defect_rate_map[cluster_id]

    top_factors = explanation_df[
        explanation_df["Cluster"] == cluster_id
    ].head(3)

    factors = top_factors["Feature"].tolist()

    print("\n" + "=" * 60)
    print("NEW PRODUCTION MONITORING RESULT")
    print("=" * 60)

    print("Production pattern:", cluster_id)
    print("Risk level:", risk_level)

    print(
        "Historical cluster defect rate:",
        round(defect_probability * 100, 2),
        "%"
    )

    print("\nPriority inspection variables:")

    for i, factor in enumerate(factors, start=1):
        print(f"{i}. {factor}")

    print("\nSuggested action:")

    if risk_level == "HIGH":
        print("Inspect associated process parameters.")
        print("Verify machine settings and sensor condition.")

    elif risk_level == "MEDIUM":
        print("Monitor process stability and verify settings.")

    else:
        print("Continue routine quality monitoring.")

    print("\nFinal decision must be made by an engineer.")

    return {
        "cluster": cluster_id,
        "risk": risk_level,
        "historical_defect_rate": defect_probability,
        "inspection_factors": factors
    }


# Select one held-out production cycle
sample_record = test_df.iloc[0][feature_cols].to_dict()

monitoring_result = monitor_new_record(sample_record)


# ============================================================
# STEP 17: PROCESS DRIFT DETECTION
#
# Compare historical production with new production.
# KS tests are applied in the original standardized space.
# Multiple-testing correction: Bonferroni.
#
# A significant change suggests process drift; it does
# not prove product defects.
# ============================================================

def detect_process_drift(reference, current, alpha=0.05):

    reference = np.asarray(reference)
    current = np.asarray(current)

    if reference.shape[1] != current.shape[1]:
        raise ValueError("Feature counts do not match.")

    n_features = reference.shape[1]

    corrected_alpha = alpha / n_features

    drift_rows = []

    for i, feature in enumerate(feature_cols):

        statistic, p_value = ks_2samp(
            reference[:, i],
            current[:, i]
        )

        drift_rows.append({
            "Feature": feature,
            "KS_Statistic": statistic,
            "P_Value": p_value,
            "Drift_Detected": p_value < corrected_alpha
        })

    drift_df = pd.DataFrame(drift_rows)

    print("\nPROCESS DRIFT RESULTS:")
    display(drift_df.sort_values("P_Value"))

    if drift_df["Drift_Detected"].any():
        print("WARNING: Significant process distribution change.")
    else:
        print("No statistically significant drift detected.")

    return drift_df


# Example: held-out dataset represents incoming observations
drift_result = detect_process_drift(
    X_train_scaled,
    X_test_scaled
)


# ============================================================
# STEP 18: DEMONSTRATE ARTIFICIAL PROCESS DRIFT
#
# This section deliberately changes one feature to
# verify that the drift checker can identify a shift.
# It is a simulation, not a real industrial event.
# ============================================================

X_shifted = X_test_scaled.copy()

# Add an artificial standardized shift to first feature
X_shifted[:, 0] += 2.0

print("\nSIMULATED DRIFT EXPERIMENT:")

simulated_drift = detect_process_drift(
    X_train_scaled,
    X_shifted
)


# ============================================================
# STEP 19: HUMAN-IN-THE-LOOP FEEDBACK
#
# Engineer decisions are stored.
# Feedback is NOT used for automatic retraining.
# A separate reviewed retraining procedure is needed.
# ============================================================

feedback_records = []

def record_engineer_feedback(
    record_id,
    cluster_id,
    decision,
    notes=""
):

    valid_decisions = [
        "Confirmed Process Issue",
        "False Alarm",
        "Sensor Problem",
        "Normal Variation",
        "Other"
    ]

    if decision not in valid_decisions:
        raise ValueError("Invalid engineer decision.")

    feedback = {
        "Record_ID": record_id,
        "Cluster": cluster_id,
        "Engineer_Decision": decision,
        "Notes": notes,
        "Recorded_At": pd.Timestamp.now().isoformat()
    }

    feedback_records.append(feedback)

    print("\nEngineer feedback saved.")

    return feedback


# Example feedback only; not a real engineer assessment
example_feedback = record_engineer_feedback(
    record_id="DEMO-001",
    cluster_id=monitoring_result["cluster"],
    decision="Other",
    notes="Demonstration record; awaiting engineer review."
)

feedback_df = pd.DataFrame(feedback_records)

print("\nHUMAN FEEDBACK LOG:")
display(feedback_df)


# ============================================================
# STEP 20: SIMPLE MANUFACTURING DECISION DASHBOARD
# ============================================================

print("\n" + "=" * 65)
print("MANUFACTURING DECISION SUPPORT DASHBOARD")
print("=" * 65)

print("\nMODEL:", best_name)
print("TRAINING SAMPLES:", len(train_df))
print("TEST SAMPLES:", len(test_df))
print("NUMBER OF CLUSTERS:", len(cluster_profile))

print("\nPRODUCTION PATTERN SUMMARY:")
display(cluster_profile)

print("\nIMPORTANT PROCESS PARAMETERS:")
display(importance_df.head(10))

print("\nCURRENT MONITORING:")
print(monitoring_result)

print("\nENGINEER VALIDATION:")
display(feedback_df)


# ============================================================
# STEP 21: SAVE MODEL AND RESULTS
# ============================================================

output_dir = "/content/manufacturing_results"

os.makedirs(output_dir, exist_ok=True)

train_df.to_csv(
    f"{output_dir}/train_cluster_results.csv",
    index=False
)

test_df.to_csv(
    f"{output_dir}/test_risk_predictions.csv",
    index=False
)

cluster_profile.to_csv(
    f"{output_dir}/cluster_risk_profile.csv"
)

explanation_df.to_csv(
    f"{output_dir}/cluster_explanations.csv",
    index=False
)

importance_df.to_csv(
    f"{output_dir}/feature_importance.csv",
    index=False
)

drift_result.to_csv(
    f"{output_dir}/drift_results.csv",
    index=False
)

feedback_df.to_csv(
    f"{output_dir}/human_feedback.csv",
    index=False
)

joblib.dump(
    {
        "imputer": imputer,
        "scaler": scaler,
        "pca": pca,
        "clustering_model": best_model,
        "surrogate_model": surrogate,
        "features": feature_cols,
        "risk_map": risk_map,
        "defect_rate_map": defect_rate_map,
        "model_name": best_name
    },
    f"{output_dir}/manufacturing_model.joblib"
)

print("\nALL RESULTS SAVED SUCCESSFULLY")
print("Location:", output_dir)


# ============================================================
# STEP 22: DOWNLOAD ALL RESULTS AS ZIP
# ============================================================

import shutil

zip_path = shutil.make_archive(
    "/content/manufacturing_results",
    "zip",
    output_dir
)

print("\nRESULTS ZIP:", zip_path)

from google.colab import files
files.download(zip_path)


# ============================================================
# PROJECT COMPLETED
# ============================================================

print("\n" + "=" * 65)
print("PROJECT EXECUTION COMPLETED")
print("=" * 65)
```
